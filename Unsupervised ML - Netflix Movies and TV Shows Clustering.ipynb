# **Project Summary -**

**Project Summary: Netflix Content Analysis and Clustering**

This capstone project provides an end-to-end machine learning analysis of the Netflix catalog as of 2019, utilizing a dataset sourced from Flixable. The primary objective is to move beyond raw data to uncover the strategic evolution of Netflix’s content library, specifically examining the shift from movies to television shows and grouping similar content through unsupervised learning. The project follows a rigorous data science pipeline, including Exploratory Data Analysis (EDA), hypothesis testing, text preprocessing, and the implementation of multiple clustering algorithms.

**Business Context and EDA**

The dataset represents a snapshot of Netflix during a period of massive global expansion. Initial exploration reveals a library of approximately 7,787 titles. A key focus of the EDA is the "Country-wise" distribution, which identifies the United States, India, and the United Kingdom as primary content producers. By analyzing the date_added and release_year columns, we can test the hypothesis that Netflix has significantly pivoted its strategy. While movies historically dominated the platform, the growth rate of TV shows has nearly tripled since 2010, confirming a strategic shift toward long-form episodic content designed to increase subscriber retention.

**Data Cleaning and Feature Engineering**

To ensure model accuracy, significant data cleaning was performed. Missing values in critical features such as director, cast, and country were handled by categorical imputation (e.g., "Unknown") rather than deletion to maintain the dataset's integrity. Feature engineering was employed to extract the "month" and "year" from date_added, allowing for the identification of seasonal content release patterns. Additionally, the duration column was standardized, separating movie runtimes (minutes) from TV show lengths (seasons) for cleaner analysis.

**Machine Learning: Text-Based Clustering**

The technical core of the project involves clustering similar content based on text-based features like description, listed_in (genres), and cast.

The preprocessing stage included converting text to lowercase, removing punctuation/stopwords, and applying Lemmatization to reduce words to their base forms. This textual data was then transformed into a numerical format using TF-IDF (Term Frequency-Inverse Document Frequency). Because the resulting matrix was high-dimensional, Principal Component Analysis (PCA) was utilized to reduce dimensionality while preserving the variance necessary for effective clustering.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Reload the dataset to ensure 'df' is in its original state before wrangling
# This is important if cells are run out of order or after previous transformations
import pandas as pd
url = '/content/drive/MyDrive/Classroom/Data Science Assignment/NETFLIX MOVIES AND TV SHOWS CLUSTERING.csv'
df = pd.read_csv(url)

# Handle missing values for date_added, director, cast, country, rating
# Fill missing categorical values with "Unknown"
categorical_columns = ['director', 'cast', 'country', 'rating']
for column in categorical_columns:
    df[column] = df[column].fillna('Unknown')

# Drop rows where 'date_added' is null as these are few and might indicate corrupted entries
# Note: This is done after filling 'rating' because 'date_added' might also have missing values
df.dropna(subset=['date_added'], inplace=True)

# Remove duplicate records
df = df.drop_duplicates()

# Convert date_added to datetime
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

# Extract year from date_added for time-based analysis
df['year_added'] = df['date_added'].dt.year

# Clean duration column by extracting only numeric values
df['duration'] = df['duration'].astype(str).str.extract(r'(\d+)', expand=False)
df['duration'] = df['duration'].fillna(0).astype(int)
````

#### Chart - 8

````python
# MONTHLY CONTENT RELEASE HEATMAP
# Chart - 8 visualization code
# 1. Preprocessing Dates
df['month_added'] = pd.to_datetime(df['date_added']).dt.month_name()
df['year_added'] = pd.to_datetime(df['date_added']).dt.year

# 2. Creating Pivot Table
# We handle missing years/months by dropping NAs
heatmap_data = df.dropna(subset=['month_added', 'year_added'])
pivot_table = heatmap_data.pivot_table(index='month_added',
                                               columns='year_added',
                                               values='show_id',
                                               aggfunc='count').fillna(0)

# Reordering months chronologically
months = ["January", "February", "March", "April", "May", "June",
                  "July", "August", "September", "October", "November", "December"]
pivot_table = pivot_table.reindex(months)

# 3. Plotting
plt.figure(figsize=(14, 8))
sns.heatmap(pivot_table, annot=True, fmt=".0f", cmap="YlGnBu", linewidths=.5)

plt.title('Heatmap: Content Added per Month (Year-wise)', fontsize=15, fontweight='bold')
plt.xlabel('Year Added', fontsize=12)
plt.ylabel('Month', fontsize=12)
plt.show()
````

##### 1. Why did you pick the specific chart?

I chose a Heatmap because it is the most effective way to represent 3-dimensional data (Year, Month, and Volume) on a 2-dimensional plane. The color intensity (from yellow to deep blue) allows for immediate pattern recognition, helping us spot "hot months" of content acquisition that a standard line graph might obscure.

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

To transition from Exploratory Data Analysis (EDA) to Inference, we will define three hypothesis statements. These are designed to validate the visual trends we observed in your 15 charts using mathematical rigor.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Statement 1: The "Attention Span" Hypothesis (Numerical)**

**Observation:** Our Histogram and Pair Plot suggested that newer movies appear shorter than older ones.


*   **Hypothesis:** There is a significant difference in the mean duration of movies released in the "Classic Era" (before 2010) compared to the "Modern Era" (2010 and later).
*   **Statistical Test:** Two-Sample T-Test (Comparing the means of two independent groups).



#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
from scipy import stats

def test_movie_duration_hypothesis(df):
    """
    Statement: Movies released after 2010 are shorter than those before 2010.
    H0: Mean duration (Pre-2010) == Mean duration (Post-2010)
    H1: Mean duration (Pre-2010) != Mean duration (Post-2010)
    """
    # 1. Data Cleaning
    movies = df[df['type'] == 'Movie'].copy()
    # The 'duration' column is already numeric (int) from data wrangling, so just convert to float
    movies['duration_min'] = movies['duration'].astype(float)
    movies = movies.dropna(subset=['duration_min'])

    # 2. Grouping
    pre_2010 = movies[movies['release_year'] < 2010]['duration_min']
    post_2010 = movies[movies['release_year'] >= 2010]['duration_min']

    # 3. Execution
    t_stat, p_val = stats.ttest_ind(pre_2010, post_2010, equal_var=False)

    print("--- Hypothesis Testing: Movie Duration ---")
    print(f"Pre-2010 Mean Duration: {pre_2010.mean():.2f} mins")
    print(f"Post-2010 Mean Duration: {post_2010.mean():.2f} mins")
    print(f"P-Value: {p_val:.4f}")

    if p_val < 0.05:
        print("Conclusion: Reject H0. There is a statistically significant difference in movie lengths.")
    else:
        print("Conclusion: Fail to reject H0. No significant difference found.")

test_movie_duration_hypothesis(df)
````

##### Which statistical test have you done to obtain P-Value?

**Independent Two-Sample T-Test**

Used for: Statement 1 (Movie Duration)


*   Why: We compared the means of two independent numerical groups (Pre-2010 movies vs. Post-2010 movies).
*   How it works: It calculates the difference between the averages of the two groups and compares it to the variation within the groups. If the difference is large enough relative to the variation, the P-Value drops below 0.05.



##### Why did you choose the specific statistical test?

**T-Test (for Movie Duration)**

The Goal: To see if the average length of movies changed significantly over time.


*   Why this test? We are comparing a numerical variable (minutes) across two categorical groups (Pre-2010 vs. Post-2010).
*   The Logic: Since movie durations generally follow a normal distribution (a bell curve where most movies are around 90–100 minutes), the T-Test is the most powerful "parametric" tool to find if the centers of those two bell curves have shifted.



### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Statement 2: The "Maturity vs. Format" Hypothesis (Categorical)**

**Observation:** Our Rating vs. Type Count Plot suggested that TV-MA (Mature) ratings might be more prevalent in TV Shows than in Movies.


*   Hypothesis: The age rating of a title is dependent on the content type (Movie vs. TV Show).
*   Statistical Test: Chi-Square Test of Independence (Testing the relationship between two categorical variables).



#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
def test_rating_type_dependency(df):
    """
    Statement: Age ratings are dependent on whether the content is a Movie or TV Show.
    H0: Rating and Type are independent.
    H1: Rating and Type are dependent.
    """
    # 1. Create Crosstab
    contingency_table = pd.crosstab(df['rating'], df['type'])

    # 2. Execution
    chi2, p_val, dof, expected = stats.chi2_contingency(contingency_table)

    print("\n--- Hypothesis Testing: Rating vs. Type Dependency ---")
    print(f"Chi-Square Statistic: {chi2:.2f}")
    print(f"P-Value: {p_val:.4e}") # Using scientific notation for very small values

    if p_val < 0.05:
        print("Conclusion: Reject H0. Content Rating is significantly dependent on Content Type.")
    else:
        print("Conclusion: Fail to reject H0. No dependency found.")

test_rating_type_dependency(df)
````

##### Which statistical test have you done to obtain P-Value?

**Chi-Square ($\chi^2$) Test of Independence**

Used for: Statement 2 (Rating vs. Content Type)


*   Why: We were comparing two categorical variables (Type: Movie/TV Show and Rating: TV-MA, PG, etc.).
*   How it works: It creates a "Contingency Table" (a grid of counts). It compares the observed frequencies in your Netflix data against the expected frequencies that would occur if there were no relationship at all. A low P-Value here means the content type definitely influences the rating.



##### Why did you choose the specific statistical test?

**Chi-Square Test (for Ratings vs. Type)**

The Goal: To see if a relationship exists between "Content Type" and "Age Rating."


*   Why this test? Both variables are categorical (e.g., Movie/TV Show and TV-MA/PG-13). You cannot calculate an "average" for a rating, so a T-test is impossible.
*   The Logic: The Chi-Square test looks at the frequency of occurrences. It asks: "If ratings were totally random, how many TV-MA shows would we expect? Does the actual count differ significantly from that expectation?"



### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Statement 3: The "Acquisition Lag" Hypothesis (Numerical)**

**Observation:** Our Lag Analysis (Chart 15) suggested that Netflix adds TV Shows to the platform faster after their release than it does for Movies.

*   Hypothesis: The average "Time-to-Platform" (Year Added minus Release Year) is significantly lower for TV Shows than for Movies.
*   Statistical Test: Mann-Whitney U Test (A non-parametric test used because "lag years" are often non-normally distributed/skewed).



#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
def test_acquisition_lag(df):
    """
    Statement: Netflix adds TV Shows faster (lower lag) than Movies.
    H0: Mean Lag (Movies) == Mean Lag (TV Shows)
    H1: Mean Lag (Movies) != Mean Lag (TV Shows)
    """
    # 1. Calculate Lag
    df['year_added'] = pd.to_datetime(df['date_added']).dt.year
    df_lag = df.dropna(subset=['year_added', 'release_year']).copy()
    df_lag['lag'] = df_lag['year_added'] - df_lag['release_year']

    # 2. Grouping
    movie_lag = df_lag[df_lag['type'] == 'Movie']['lag']
    tv_lag = df_lag[df_lag['type'] == 'TV Show']['lag']

    # 3. Execution (Non-parametric as lag is skewed)
    u_stat, p_val = stats.mannwhitneyu(movie_lag, tv_lag)

    print("\n--- Hypothesis Testing: Acquisition Lag ---")
    print(f"Median Movie Lag: {movie_lag.median()} years")
    print(f"Median TV Show Lag: {tv_lag.median()} years")
    print(f"P-Value: {p_val:.4e}")

    if p_val < 0.05:
        print("Conclusion: Reject H0. There is a significant difference in acquisition speed.")
    else:
        print("Conclusion: Fail to reject H0. No significant difference found.")

test_acquisition_lag(df)
````

##### Which statistical test have you done to obtain P-Value?

**Mann-Whitney U Test**

Used for: Statement 3 (Acquisition Lag)


*   Why: This is a non-parametric test. Our "Lag" data (Year Added minus Release Year) is highly skewed—most content has 0-1 years of lag, but some has 50 years. Because the data isn't a perfect "bell curve," a standard T-Test would be inaccurate.
*   How it works: Instead of comparing means (averages), it compares the ranks of the data points. It is much more robust for data with outliers or non-normal distributions.



##### Why did you choose the specific statistical test?

**Mann-Whitney U Test (for Acquisition Lag)**

The Goal: To see if TV Shows are added to Netflix faster than Movies.


*   Why this test? The "Lag" data (Year Added minus Release Year) is non-normal and skewed. Most values are 0 or 1, but some are 50 (old classics).
*   The Logic: A standard T-test is easily "tricked" by outliers (extreme values). The Mann-Whitney U test is non-parametric, meaning it doesn't care about the bell curve. Instead, it ranks the data points. If TV Shows consistently rank "faster" than Movies, it proves the difference is real without being skewed by a few 50-year-old movies.

````python
# ML Model - 1 Implementation

# Fit the Algorithm
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Initialize TF-IDF Vectorizer
# We limit to 5000 features to keep the model efficient
tfidf = TfidfVectorizer(stop_words='english', max_features=5000)

# 2. Construct the TF-IDF matrix (Fitting)
# 'metadata_soup' is the column containing genres, cast, and description
tfidf_matrix = tfidf.fit_transform(train_df['metadata_soup'])

# 3. Calculate Cosine Similarity scores
# This creates a square matrix where [i, j] is the similarity between movie i and j
cosine_sim_matrix = cosine_similarity(tfidf_matrix, tfidf_matrix)

print(f"Model fitted. Matrix shape: {cosine_sim_matrix.shape}")
# Predict on the model
def get_recommendations(title, cosine_sim=cosine_sim_matrix):
    # 1. Get the index of the movie that matches the title
    idx = train_df.index[train_df['title'] == title].tolist()[0]

    # 2. Get the pairwise similarity scores of all movies with that movie
    sim_scores = list(enumerate(cosine_sim[idx]))

    # 3. Sort the movies based on the similarity scores
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # 4. Get the scores of the 10 most similar movies (skipping the first one as it is itself)
    sim_scores = sim_scores[1:11]

    # 5. Get the movie indices
    movie_indices = [i[0] for i in sim_scores]

    # 6. Return the top 10 most similar movies
    return train_df['title'].iloc[movie_indices]

# Test the prediction
print("Recommendations for 'Stranger Things':")
print(get_recommendations('Stranger Things'))
````

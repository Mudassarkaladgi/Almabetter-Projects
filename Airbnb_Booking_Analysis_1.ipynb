## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Fill missing values in 'reviews_per_month' with 0
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)

# Fill missing values in 'last_review' with a placeholder (e.g., 'No reviews')
df['last_review'] = df['last_review'].fillna('No reviews')

# Check if missing values are handled
print(df.isnull().sum())
````

### What all manipulations have you done and insights you found?

Data Manipulations

The data wrangling process was focused on cleaning and preparing the raw data to ensure the reliability of our analysis. The key manipulations were:

- Handling Missing Values:

  -  Missing values in the reviews_per_month column were filled with 0. This is a logical assumption, as a missing value in this context likely means the listing has no reviews yet.

  - Missing values in the last_review, name, and host_name columns were replaced with "Not Available." This preserves the data's structure without removing potentially useful listings from the dataset.

- Removing Duplicate Rows:

  - The code checked for and removed any exact duplicate rows. This is a crucial step to ensure each entry represents a unique listing and to avoid skewing our analysis.

- Handling Outliers:

  - Listings with a price of 0 were removed. A price of zero is not a valid rental price and would distort any analysis of pricing trends.

  - Listings with an extremely high minimum_nights value (specifically, greater than 365) were filtered out. These are likely not standard short-term rentals and would skew the analysis of booking behavior.

After these manipulations, the dataset is much cleaner and more suitable for reliable analysis and visualization.

Initial Insights

Even before creating visualizations, the data's structure and contents provide some immediate insights:

- Geographical Concentration: The dataset is heavily concentrated in a few key neighbourhood groups (boroughs) in New York City, such as Manhattan and Brooklyn, with very few listings in other areas like Staten Island. This tells us where Airbnb's market is most mature.

- Market Segmentation by Room Type: The room_type variable is a critical factor in the dataset, with three distinct categories: "Entire home/apt," "Private room," and "Shared room." This is the primary way guests choose a listing, so it will be a key variable for understanding pricing differences.

- A Mix of Host Types: The calculated_host_listings_count column has a wide range of values. This indicates a mix of host types, from individuals with a single listing to professional hosts with dozens or even hundreds of listings.

- Review Activity: There's a notable number of listings with zero reviews, as confirmed by the missing values in the reviews_per_month column. This suggests that a portion of the listings in the dataset are either new or have not received any bookings.

## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

#### Chart - 1

````python
# Chart - 1 visualization code
# Group data by availability and calculate the average price
avg_price_by_availability = df.groupby('availability_365')['price'].mean().reset_index()

# --- Visualization ---
plt.figure(figsize=(14, 8))
sns.lineplot(data=avg_price_by_availability, x='availability_365', y='price', marker='o')

plt.title('Average Price vs. Availability (Days in a Year)', fontsize=16)
plt.xlabel('Availability (Days)', fontsize=12)
plt.ylabel('Average Price ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
````

##### 1. Why did you pick the specific chart?

I chose the line chart because it's the best way to visualize how one continuous number (price) changes as another continuous number (availability) changes. It clearly shows the overall trend, making it easy to spot relationships and tell a story about the data without getting lost in individual data points.

##### 2. What is/are the insight(s) found from the chart?

The insight is that properties with lower availability tend to have a higher price. This suggests that more expensive listings are likely unique or luxury properties that are rented out less frequently, while more affordable listings are available more often.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Based on the insights:

- Positive Impact: Airbnb can help hosts with lower-priced, high-availability listings earn more by encouraging dynamic pricing for peak demand. This helps both the host and Airbnb's revenue.
